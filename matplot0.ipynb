```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
```

```python
df['species']=iris.target
df['species'] = df['species'].map({0:"setosa", 1:"vesicolar", 2:"virginica"})
```

```python
print(df)
```

```python
plt.plot(df['sepal length (cm)'], "r--")
```

```python
df.plot(kind="scatter", x='sepal length (cm)', y='petal length (cm)')
```

```python
df.plot(kind="scatter", x='sepal length (cm)', y='petal length (cm)')
plt.grid()
```

```python
import seaborn as sns
```

```python
sns.FacetGrid(df, hue="species", height=6)
```

```python
sns.FacetGrid(df, hue="species", height=6) .map(plt.scatter, 'sepal length (cm)', 'petal length (cm)')
```

```python
sns.FacetGrid(df, hue="species", height=6) .map(plt.scatter, 'sepal length (cm)', 'petal length (cm)') .add_legend()
```

```python
_,axes=plt.subplots(1,2, sharey=True, figsize=(10,4))
```

```python
_,axes=plt.subplots(1,2, sharey=True, figsize=(10,4))
sns.boxplot(x='petal length (cm)', y='species', data=df, ax=axes[0])
sns.violinplot(x='petal length (cm)', y='species', data=df, ax=axes[1])
```
